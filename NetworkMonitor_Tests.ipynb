## Run Unit Tests

Let's start by running the basic unit tests for the NetworkMonitor class.

```python
# Create a test suite for NetworkMonitor tests
def run_test_suite(test_module):
    # Reload the module to ensure we have the latest version
    importlib.reload(test_module)
    
    # Get all test cases from the module
    loader = unittest.TestLoader()
    suite = loader.loadTestsFromModule(test_module)
    
    # Run the tests using TextTestRunner with verbose output
    runner = unittest.TextTestRunner(verbosity=2)
    result = runner.run(suite)
    
    # Return test results
    return {
        "tests_run": result.testsRun,
        "failures": len(result.failures),
        "errors": len(result.errors),
        "skipped": len(result.skipped),
        "was_successful": result.wasSuccessful()
    }

# Run the NetworkMonitor unit tests
print("Running NetworkMonitor Unit Tests...")
network_monitor_results = run_test_suite(test_network_monitor)
print(f"\nTest Results: {network_monitor_results}")
```

## Run Integration Tests

Now let's run the integration tests.

```python
# Run the NetworkMonitor integration tests
print("Running NetworkMonitor Integration Tests...")
integration_results = run_test_suite(test_network_monitor_integration)
print(f"\nTest Results: {integration_results}")
```

## Run Network Traffic Processing Tests

These tests verify the network traffic processing functionality.

```python
# Run the network traffic processing tests
print("Running Network Traffic Processing Tests...")
traffic_processing_results = run_test_suite(test_network_traffic_processing)
print(f"\nTest Results: {traffic_processing_results}")
```

## Run Response Action Tests

These tests verify the IP blocking and unblocking functionality.

```python
# Run the response action tests
print("Running Response Action Tests...")
response_action_results = run_test_suite(test_response_actions)
print(f"\nTest Results: {response_action_results}")
```

## Run Threat Detection Tests

These tests verify the threat detection capabilities.

```python
# Run the threat detection tests
print("Running Threat Detection Tests...")
threat_detection_results = run_test_suite(test_threat_detection)
print(f"\nTest Results: {threat_detection_results}")
```

## Test Results Summary

Let's visualize the results of all test suites.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Collect results from all test suites
test_results = {
    "NetworkMonitor Unit Tests": network_monitor_results,
    "Integration Tests": integration_results,
    "Traffic Processing Tests": traffic_processing_results,
    "Response Action Tests": response_action_results,
    "Threat Detection Tests": threat_detection_results
}

# Create DataFrame for visualization
results_df = pd.DataFrame({
    "Test Suite": list(test_results.keys()),
    "Tests Run": [result["tests_run"] for result in test_results.values()],
    "Passed": [result["tests_run"] - result["failures"] - result["errors"] - result["skipped"] for result in test_results.values()],
    "Failed": [result["failures"] for result in test_results.values()],
    "Errors": [result["errors"] for result in test_results.values()],
    "Skipped": [result["skipped"] for result in test_results.values()],
    "Success Rate": [
        100 * (result["tests_run"] - result["failures"] - result["errors"] - result["skipped"]) / result["tests_run"] 
        if result["tests_run"] > 0 else 0 
        for result in test_results.values()
    ]
})

# Display the results table
display(results_df)

# Plot the results
plt.figure(figsize=(12, 6))
results_df.plot(x="Test Suite", y=["Tests Run", "Passed", "Failed", "Errors", "Skipped"], kind="bar")
plt.title("Test Suite Results")
plt.ylabel("Number of Tests")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# Plot success rate
plt.figure(figsize=(10, 5))
results_df.plot(x="Test Suite", y="Success Rate", kind="bar", color="green")
plt.title("Test Success Rate")
plt.ylabel("Success Rate (%)")
plt.xticks(rotation=45, ha="right")
plt.ylim(0, 105)  # Set y-axis from 0 to 105%
plt.axhline(y=100, color='r', linestyle='-', alpha=0.3)  # Add a red line at 100%
plt.tight_layout()
plt.show()
```

## Run Specific Test Cases

You can also run specific test cases by name. This is useful for debugging specific issues.

```python
def run_specific_test(test_module, test_case_name):
    # Reload the module to ensure we have the latest version
    importlib.reload(test_module)
    
    # Get all test cases from the module
    loader = unittest.TestLoader()
    
    # Try to find the specific test case
    try:
        suite = loader.loadTestsFromName(f"{test_module.__name__}.{test_case_name}")
        
        # Run the tests using TextTestRunner with verbose output
        runner = unittest.TextTestRunner(verbosity=2)
        result = runner.run(suite)
        
        return result.wasSuccessful()
    except AttributeError:
        print(f"Test case '{test_case_name}' not found in module {test_module.__name__}")
        return False

# Example: Run a specific test case
# Change these values to run different tests
module_to_test = test_network_monitor
test_case_to_run = "TestNetworkMonitor.test_check_threat_indicators_ip"

print(f"Running specific test: {module_to_test.__name__}.{test_case_to_run}")
success = run_specific_test(module_to_test, test_case_to_run)
print(f"Test {'passed' if success else 'failed'}")
```
